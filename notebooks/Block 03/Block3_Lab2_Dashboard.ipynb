The operations manager of an online shop asks three questions every Monday:

1. **How many orders arrive late?** One number, with its denominator.
2. **Is it getting better or worse?** The late rate month by month.
3. **Where is it worst?** The late rate by customer state, with the number of orders behind each rate.

You build the answers in five tasks:

| Task | You build | The AI's role |
|---|---|---|
| 1 | The three headline numbers | Writes the code |
| 2 and 3 | Two charts | Writes the code |
| 4 | A written summary | **Writes the words**, from facts you give it, under rules you set; a check tests every number |
| 5 | A dashboard with a state filter and an **Explain this view** button | **Sits inside the product**: it explains whichever state the manager picks, and the check runs every time |

In each task: **plan**, **prompt** (Context, Task, Constraints, Format, Check), **read** the code, **run** it, **verify** one number, **explain** it for the manager.

**The rule for LLM text: facts in, words out, numbers checked.** The LLM never sees the raw data, only a short list of facts you computed. Any number it writes that is not in those facts is flagged.
